# Dataset Loading and Initial Exploration

Before performing the required analyses, all CSV files are loaded and inspected to understand the structure of the dataset.

The dataset contains trending YouTube videos from multiple countries. Each CSV file represents a different country and will later be combined into a single dataframe.

```python
csv_files = glob.glob("data/*videos.csv.zst")

print("Number of CSV files found:", len(csv_files))

for file in csv_files:
    print(file)
```

# Question 1

Create a single dataframe with the concatenation of all input CSV files, adding a column called country.

Each CSV file corresponds to one country. The country code is extracted from the first two letters of the filename, for example `USvideos.csv.zst` gives `US`.

```python
dataframes = []

for file in csv_files:
    country = os.path.basename(file)[:2]
    
    df = pd.read_csv(
        file,
        compression="zstd",
        encoding="latin1"
    )
    
    df["country"] = country
    dataframes.append(df)

videos = pd.concat(dataframes, ignore_index=True)

print("Total rows:", videos.shape[0])
print("Total columns:", videos.shape[1])

videos.head()
```

```python
videos["country"].value_counts()
```

# Question 2

## Extract all videos that have no tag

The dataset uses the value `[none]` to represent videos without tags.

To answer this question, a new dataframe called `no_tags` is created by filtering the rows where the `tags` column contains the value `[none]`.

```python
no_tags = videos[videos["tags"] == "[none]"]

print("Number of videos without tags:", len(no_tags))

no_tags.head()
```

```python
no_tags[["video_id", "title", "channel_title", "country"]].head(10)
```

# Question 3

## Total number of views for each channel

To answer this question, videos are grouped by channel title and the total number of views is computed using the sum aggregation function.

This allows us to determine which channels accumulated the highest number of views in the dataset.

```python
views_per_channel = (
    videos.groupby("channel_title")["views"]
    .sum()
    .reset_index()
)

views_per_channel = views_per_channel.rename(
    columns={"views": "total_views"}
)

views_per_channel.head()
```

```python
views_per_channel = views_per_channel.sort_values(
    by="total_views",
    ascending=False
)

views_per_channel.head(10)
```

### Observations

The channels with the highest total number of views are dominated by music and entertainment content. ChildishGambinoVEVO and Marvel Entertainment accumulated more than 10 billion views across all their trending videos in the dataset.

This suggests that music videos, entertainment content, and viral productions tend to attract significantly more views than other categories on YouTube.

# Question 4

## Create the excluded dataframe

Videos with disabled comments, disabled ratings, or videos marked as removed are stored in a separate dataframe called `excluded`.

These rows are then removed from the main dataframe to ensure that subsequent analyses are performed only on valid videos.

```python
excluded = videos[
    (videos["comments_disabled"] == True) |
    (videos["ratings_disabled"] == True) |
    (videos["video_error_or_removed"] == True)
]

print("Excluded videos:", len(excluded))
```

```python
videos = videos.drop(excluded.index)

print("Remaining videos:", len(videos))
```

```python
print("Excluded videos:", len(excluded))
print("Remaining videos:", len(videos))
print("Total:", len(excluded) + len(videos))
```

### Observations

A subset of videos had disabled comments, disabled ratings, or were marked as removed. These videos were separated into the `excluded` dataframe and removed from the main dataset to avoid affecting the analyses performed in the following questions.

# Question 5

## Create the like_ratio column

The like ratio measures the relationship between the number of likes and dislikes for a video.

It is computed as:

like_ratio = likes / dislikes

Videos with zero dislikes are excluded from the calculation to avoid division by zero.

```python
(videos["dislikes"] == 0).sum()
```

```python
videos["like_ratio"] = (
    videos["likes"] /
    videos["dislikes"].replace(0, np.nan)
)
```

```python
videos[["likes", "dislikes", "like_ratio"]].head(10)
```

```python
videos["like_ratio"].describe()
```

### Observations

The `like_ratio` column was added to measure the relationship between likes and dislikes for each video.

Videos with zero dislikes cannot produce a valid ratio because division by zero is undefined. These cases were assigned a missing value (`NaN`) to preserve the integrity of the dataset.

# Question 6

## Cluster publish times into 10-minute intervals

The `publish_time` column contains full timestamps including date and time.

To analyze publication patterns, each timestamp is converted to a 10-minute interval. Videos published between the same 10-minute boundaries are assigned to the same interval.

```python
videos["publish_time"] = pd.to_datetime(videos["publish_time"])
```

```python
videos["publish_interval"] = (
    videos["publish_time"]
    .dt.floor("10min")
    .dt.strftime("%H:%M")
)
```

```python
videos[
    ["publish_time", "publish_interval"]
].head(10)
```

### Observations

The publication timestamps were grouped into 10-minute intervals. This transformation simplifies time-based analysis and allows trends to be studied according to the time of publication rather than exact timestamps.

# Question 7

## Statistics for each 10-minute interval

Videos are grouped according to their publication interval.

For each interval, the following statistics are computed:

- Number of videos
- Average number of likes
- Average number of dislikes

This helps identify whether certain publication periods are associated with higher engagement.

```python
interval_stats = (
    videos.groupby("publish_interval")
    .agg(
        video_count=("video_id", "count"),
        avg_likes=("likes", "mean"),
        avg_dislikes=("dislikes", "mean")
    )
    .reset_index()
)
```

```python
interval_stats = interval_stats.sort_values(
    by="publish_interval"
)

interval_stats.head(15)
```

### Observations

The number of videos published varies across the different 10-minute intervals.

Some intervals contain considerably more videos than others. In addition, average likes and dislikes fluctuate throughout the day, suggesting that publication time may influence audience engagement and video visibility.

# Question 8

## Number of videos for each tag

The `tags` column contains multiple tags separated by the `|` character.

To determine the number of videos associated with each tag, the tags are separated into individual values and then counted across the entire dataset.

```python
tags_df = videos[["video_id", "tags"]].copy()
```

```python
tags_df["tags"] = tags_df["tags"].str.split("|")
```

```python
tags_df = tags_df.explode("tags")
```

```python
tags_df = tags_df[
    (tags_df["tags"] != "[none]") &
    (tags_df["tags"].notna())
]
```

```python
tag_counts = (
    tags_df.groupby("tags")
    .size()
    .reset_index(name="video_count")
)
```

```python
tag_counts = tag_counts.sort_values(
    by="video_count",
    ascending=False
)

tag_counts.head(20)
```

# Question 9

## Tags with the largest number of videos

Using the tag counts computed in Question 8, the tags are sorted in descending order to identify the tags that appear in the largest number of videos.

```python
top_tags = (
    tag_counts
    .sort_values(
        by="video_count",
        ascending=False
    )
    .head(10)
)

top_tags
```

### Observations

The most common tags are mainly related to entertainment, humor, music, and news content.

The tag "funny" appears in the largest number of videos (14,834 videos), followed by "comedy" (11,900 videos) and "2018" (10,567 videos).

This suggests that entertainment-focused content dominates the trending videos dataset.

# Question 10

## Average like/dislike ratio for each (tag, country) pair

The goal is to compute the average value of the `like_ratio` column for every combination of tag and country.

To achieve this, the exploded tags dataframe is combined with the country and like_ratio information from the main dataset. The results are then grouped by tag and country and the average ratio is calculated.

```python
tag_country_ratio = (
    tags_df.merge(
        videos[["video_id", "country", "like_ratio"]],
        on="video_id"
    )
)
```

```python
tag_country_ratio = (
    tag_country_ratio
    .groupby(["tags", "country"])["like_ratio"]
    .mean()
    .reset_index(name="avg_like_ratio")
)

tag_country_ratio.head(20)
```

```python
tag_country_ratio.sort_values(
    by="avg_like_ratio",
    ascending=False
).head(20)
```

### Observations

The average like ratio varies considerably across different tags and countries.

Some tag-country combinations have extremely high ratios because the associated videos received very few dislikes compared to the number of likes.

This analysis highlights differences in audience reception for specific content categories across countries.

# Question 12

## Split trending date into year, month, and day

The `trending_date` column stores the date in the format `YY.DD.MM`.

To facilitate further analysis, the column is separated into three independent columns: `year`, `month`, and `day`.

```python
date_parts = videos["trending_date"].str.split(".", expand=True)

videos["year"] = "20" + date_parts[0]
videos["day"] = date_parts[1]
videos["month"] = date_parts[2]
```

```python
videos["year"] = videos["year"].astype(int)
videos["month"] = videos["month"].astype(int)
videos["day"] = videos["day"].astype(int)
```

```python
videos[
    ["trending_date", "year", "month", "day"]
].head(10)
```

### Observations

The `trending_date` column was successfully decomposed into separate year, month, and day columns.

This transformation simplifies date-based analyses and will be used in the next question to identify the most viewed videos for each month and country.

# Question 14

## Read all JSON files with the video categories

Each country has a JSON file containing the mapping between `category_id` and the category title.

The goal of this question is to read all category JSON files and create a dataframe containing the country, category ID, and category title.

```python
json_files = glob.glob("data/*_category_id.json")

print("Number of JSON files found:", len(json_files))

for file in json_files:
    print(file)
```

```python
categories = []

for file in json_files:
    country = os.path.basename(file)[:2]
    
    with open(file, "r", encoding="utf-8") as f:
        data = json.load(f)
    
    for item in data["items"]:
        categories.append({
            "country": country,
            "category_id": int(item["id"]),
            "category_title": item["snippet"]["title"]
        })

categories_df = pd.DataFrame(categories)

categories_df.head(20)
```

```python
categories_df["country"].value_counts()
```

### Observations

All category JSON files were successfully loaded and combined into the `categories_df` dataframe.

The dataset contains category mappings for all 10 countries. Most countries have 31 video categories, while the United States contains 32 categories. These mappings will be useful for associating category IDs with readable category names in subsequent analyses.

# Question 15

## Videos with non-assignable categories by country

A video category is considered not assignable when its `category_id` does not have a matching category title in the category JSON file of the same country.

To identify these videos, the main videos dataframe is merged with the category dataframe using both `country` and `category_id`.

```python
videos_with_categories = videos.merge(
    categories_df,
    on=["country", "category_id"],
    how="left"
)

videos_with_categories.head()
```

```python
not_assignable = videos_with_categories[
    videos_with_categories["category_title"].isna()
]

not_assignable.head()
```

```python
not_assignable_by_country = (
    not_assignable.groupby("country")
    .size()
    .reset_index(name="not_assignable_count")
)

not_assignable_by_country
```

```python
not_assignable_by_country.sort_values(
    by="not_assignable_count",
    ascending=False
)
```

### Observations

The number of videos with non-assignable categories varies significantly across countries.

Russia (RU) has the highest number of videos with categories that could not be assigned (1301 videos), followed by South Korea (KR) and Germany (DE).

These videos contain category IDs that do not have a corresponding category definition in the category JSON files. Such inconsistencies may be caused by missing category mappings, changes in YouTube category definitions, or country-specific categories not present in the provided metadata.

Japan (JP) has the fewest non-assignable videos, indicating that nearly all category IDs in the dataset were successfully matched to a category title.
